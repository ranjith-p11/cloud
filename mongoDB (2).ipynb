```python
db = client["student"]
records = db["student_table"]

print("Database:", db.name)
print("Collection:", records.name)
```

```python
new_student = {
    "name": "Arun",
    "mail": "arun@example.com",
    "phone": 101
}

result = records.insert_one(new_student)

print("One record inserted successfully")
print("Inserted ID:", result.inserted_id)
```

```python
new_students = [
    {
        "name": "Banu",
        "mail": "banu@example.com",
        "phone": 102
    },
    {
        "name": "Charan",
        "mail": "charan@example.com",
        "phone": 103
    }
]

result = records.insert_many(new_students)

print("Multiple records inserted successfully")
print("Inserted IDs:", result.inserted_ids)
```

```python
for student in records.find():
    print(student)
```

```python
student = records.find_one({"phone": 102})

print(student)
```

```python
result = records.update_one(
    {"phone": 102},
    {"$set": {"mail": "banu_new@example.com"}}
)

print("Matched:", result.matched_count)
print("Modified:", result.modified_count)

print(records.find_one({"phone": 102}))
```

```python
result = records.delete_one({"phone": 103})

print("Deleted records:", result.deleted_count)
```
